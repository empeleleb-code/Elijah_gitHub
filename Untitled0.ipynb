### Q1(a) Data Profile and Semantic Hazards

**Transaction Count vs. Unique Account Count**
Transaction-level uniqueness (e.g., `REFNUM`) does not establish person-level uniqueness. A single account can generate multiple transactions. Therefore, counting rows overcounts the privacy subjects. We must use `ACCT_NO` to establish the true privacy unit.

```python
# Cell 2: Data Profiling
print("=== DATA PROFILE ===")
total_rows = len(df)
unique_refs = df['REFNUM'].nunique()
unique_accts = df['ACCT_NO'].nunique()
unique_hpans = df['HPAN'].nunique()

print(f"Total Rows (Transactions): {total_rows}")
print(f"Unique REFNUM (Transaction IDs): {unique_refs}")
print(f"Unique ACCT_NO (Privacy Subjects): {unique_accts}")
print(f"Unique HPAN (Masked Cards): {unique_hpans}")

# Check for constant columns and high-null columns
constant_cols = [col for col in df.columns if df[col].nunique() <= 1]
high_null_cols = [col for col in df.columns if df[col].isnull().mean() > 0.5]

print(f"\nConstant Columns ({len(constant_cols)}): {constant_cols[:5]}...")
print(f"High-Null Columns (>50% missing) ({len(high_null_cols)}): {high_null_cols[:5]}...")

# Identify potential duplicate columns
dup_candidates = []
if df['ACCT_NO'].equals(df['ACCT1']):
    dup_candidates.append("ACCT_NO and ACCT1 are identical")

print(f"\nDuplicate column check: {dup_candidates}")

# Check response and status distributions
print("\n=== RESPONSE/STATUS DISTRIBUTIONS ===")
print(df['TRANS_STAT'].value_counts())
print("\n", df['RESPCODE'].value_counts().head())
```

**Six Data-Quality or Semantic Hazards Identified:**
1. **Identifier Linkability:** `HPAN` is masked but deterministic (e.g., `999999XXXXXX0157`). It acts as a direct link to the card.
2. **Repeated Direct Identifiers:** `ACCT_NO` and `ACCT1` repeat across multiple rows, proving that transaction-level counts (500) will overcount the actual privacy subjects.
3. **Semantic Ambiguity (Outcomes):** `TRANS_STAT` contains "DECLINED" and "REVERSAL" alongside "SUCCESS". A declined transaction does not imply fraud; it may be insufficient funds or a timeout.
4. **Constant/Empty Columns:** Fields like `ACCT2` and `ACCT2TP` are entirely empty, providing no analytical value but expanding the attack surface (149 columns).
5. **Quasi-Identifier Exposure:** `ADDRESS_CITY` (e.g., Durban, Polokwane) combined with timestamps creates a high risk of linkage attacks.
6. **Negative Balances / Zero Values:** `ACCTBAL` contains negative values (e.g., -54887), which could distort aggregate calculations if not clipped or validated.

### Q1(b) End-to-End Architecture

**Trust Boundaries and Flow:**
1. **Ingestion & Quarantine:** Raw CSV arrives. Validated against `data_contract.json`. Malformed records are moved to a restricted quarantine zone (never deleted).
2. **Bronze (Raw Minimised):** Direct identifiers (`REFNUM`, `ACCT_NO`, `HPAN`) are extracted and sent to a Token Vault. Only the HMAC-SHA256 token is kept in the Bronze layer.
3. **Silver (Analytics):** Generalised QIs (City -> Region) and hashed tokens. Fraud operations can access this via ABAC (Purpose="Fraud").
4. **Gold (Aggregates):** Pre-calculated metrics (e.g., daily decline rates by region). This is the ONLY layer exposed to external partners.
5. **Controlled Release:** External partner downloads via an API with strict token expiry.

**AWS & Azure Service Mapping (Proposed, Not Deployed):**
| Function | AWS Service | Azure Service |
| :--- | :--- | :--- |
| **Storage** | Amazon S3 (Bronze/Silver/Gold buckets) | Azure Data Lake Storage Gen2 |
| **IAM / Access** | AWS IAM + AWS Lake Formation | Microsoft Entra ID + Azure RBAC |
| **Key Management** | AWS KMS (for HMAC & AES keys) | Azure Key Vault |
| **Audit Logging** | AWS CloudTrail | Azure Monitor / Log Analytics |

**Three Rejected Alternatives:**
1. *Proposal:* Upload all 149 fields to a common analytics lake. *Rejection Reason:* Violates data minimisation. The 149 columns contain unnecessary PII and empty fields.
2. *Proposal:* Email a filtered CSV to partners. *Rejection Reason:* No revocation. Once emailed, the data cannot be audited or controlled.
3. *Alternative:* Use base64 encoding to "mask" account numbers. *Rejection Reason:* Base64 is reversible encoding, not encryption or true anonymisation.

### Q1(d) Architecture Decision Record (ADR)

**Title:** Resolving Analyst Utility vs. Privacy Minimisation
**Context:** Fraud analysts requested access to raw `TRANS_STAT` and `ACCT_NO` to investigate declined transactions. Privacy engineering mandates minimisation.
**Decision:** Analysts will NOT access raw `ACCT_NO`. Instead, they receive an HMAC-hashed account token. Analysts can link multiple transactions to the same token but cannot reverse it to the real account.
**Consequences:** Analysts can still track fraud rings by token, but require a secure API lookup (requiring justification) if they need the actual account number for a case.

### Q2(a) HMAC Tokenization and Minimised Views
We implement keyed HMAC-SHA256 to replace `ACCT_NO`. The key must not be hardcoded in production; here, we simulate retrieving it from an environment variable or KMS.

```python
import hmac
import hashlib
import os

# Simulated secret key from KMS (Do not hardcode in production!)
HMAC_SECRET_KEY = b"super_secret_kms_key_123"

def tokenize_account(acct_no):
    if pd.isna(acct_no):
        return None
    return hmac.new(HMAC_SECRET_KEY, str(acct_no).encode('utf-8'), hashlib.sha256).hexdigest()

# Create minimised analytics view
df_minimised = df_clean.copy()
df_minimised['ACCT_TOKEN'] = df_minimised['ACCT_NO'].apply(tokenize_account)

# Drop direct identifiers to create the view
df_analytics_view = df_minimised.drop(columns=['ACCT_NO', 'ACCT1', 'HPAN', 'REFNUM'])

# EXECUTABLE ASSERTIONS
assert 'ACCT_NO' not in df_analytics_view.columns, "Direct identifier ACCT_NO was not removed!"
# Verify deterministic join (same account = same token)
assert df_analytics_view['ACCT_TOKEN'].nunique() == df_clean['ACCT_NO'].nunique(), "Tokenization lost uniqueness!"
print("[PASS] HMAC tokenization successful. Direct identifiers removed from analytics view.")
```

### Q2(b) AES-GCM Envelope Encryption
We implement envelope encryption using AES-256-GCM. We use a DEK (Data Encryption Key) to encrypt the payload and a KEK (Key Encryption Key) to wrap the DEK. We will test for tampering, wrong context (AAD), and KEK rotation.

```python
from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes
from Crypto.Protocol.KDF import HKDF

# Simulated KEKs (Version 1 and Version 2)
KEK_V1 = get_random_bytes(32)
KEK_V2 = get_random_bytes(32)

def encrypt_payload(payload, kek, context_aad):
    dek = get_random_bytes(32) # Fresh DEK
    cipher = AES.new(dek, AES.MODE_GCM)
    cipher.update(context_aad.encode('utf-8')) # Associated Data
    ciphertext, tag = cipher.encrypt_and_digest(payload.encode('utf-8'))

    # Wrap the DEK using the KEK (In production, use AES-KWP, here we simulate with XOR)
    wrapped_dek = bytes([a ^ b for a, b in zip(dek, kek)])
    return cipher.nonce, ciphertext, tag, wrapped_dek

def decrypt_payload(nonce, ciphertext, tag, wrapped_dek, kek, context_aad):
    dek = bytes([a ^ b for a, b in zip(wrapped_dek, kek)]) # Unwrap
    cipher = AES.new(dek, AES.MODE_GCM, nonce=nonce)
    cipher.update(context_aad.encode('utf-8'))
    return cipher.decrypt_and_verify(ciphertext, tag).decode('utf-8')

payload = "Sensitive_Transaction_Data_50000"
aad = "purpose=fraud_ops"

# 1. Encrypt with KEK_V1
nonce, ciphertext, tag, wrapped_dek = encrypt_payload(payload, KEK_V1, aad)

# 2. Test Correct Decryption
assert decrypt_payload(nonce, ciphertext, tag, wrapped_dek, KEK_V1, aad) == payload
print("[PASS] Correct decryption successful.")

# 3. Test Tamper Failure
tampered_ciphertext = bytearray(ciphertext)
tampered_ciphertext[0] ^= 1 # Flip one bit
try:
    decrypt_payload(nonce, bytes(tampered_ciphertext), tag, wrapped_dek, KEK_V1, aad)
    assert False, "Tamper test failed to throw error."
except ValueError:
    print("[PASS] Adversarial test: Ciphertext tampering detected.")

# 4. Test Wrong Context Failure
try:
    decrypt_payload(nonce, ciphertext, tag, wrapped_dek, KEK_V1, "purpose=reporting")
    assert False, "Wrong context failed to throw error."
except ValueError:
    print("[PASS] Adversarial test: Wrong context (AAD) detected.")

# 5. KEK Rotation (Re-wrapping DEK without rewriting payload)
# Unwrap with V1, then wrap with V2
unwrapped_dek = bytes([a ^ b for a, b in zip(wrapped_dek, KEK_V1)])
new_wrapped_dek = bytes([a ^ b for a, b in zip(unwrapped_dek, KEK_V2)])

# Verify access with old ciphertext but new KEK
assert decrypt_payload(nonce, ciphertext, tag, new_wrapped_dek, KEK_V2, aad) == payload
print("[PASS] KEK rotation successful without rewriting encrypted payload.")
```

### Q2(c) RBAC + ABAC Enforced Access
We define an access control function combining Role-Based Access Control (RBAC) and Attribute-Based Access Control (ABAC).

```python
def check_access(role, purpose, classification, expiry_hours):
    # ABAC conditions
    if expiry_hours <= 0:
        return "DENY: Token expired"
    if purpose not in ["fraud", "reporting"]:
        return "DENY: Unapproved purpose"

    # RBAC + ABAC Matrix
    if role == "analyst":
        if classification == "raw":
            return "DENY: Analysts cannot see raw data"
        return "ALLOW: Analyst access to aggregates/minimised"

    elif role == "fraud_investigator":
        if purpose == "fraud" and classification == "minimised":
            return "ALLOW: Fraud access to minimised data"
        return "DENY: Fraud investigator requires fraud purpose & minimised classification"

    elif role == "external_partner":
        if classification == "aggregate":
            return "ALLOW: Partner access to aggregates"
        return "DENY: Partners can only see aggregates"

    return "DENY: Unknown role"

# Tests
assert "DENY" in check_access("analyst", "reporting", "raw", 24)
assert "ALLOW" in check_access("fraud_investigator", "fraud", "minimised", 24)
assert "DENY" in check_access("external_partner", "reporting", "aggregate", -1) # Expired
print("[PASS] RBAC/ABAC access matrix enforced correctly.")
```

### Q2(d) DLP Rules
We build regex rules for account numbers and card fragments, evaluate them with a small test corpus, and demonstrate an evasion.

```python
import re

# Rule 1: 16-digit card with masked X's
dlp_card_rule = re.compile(r"\b\d{6}X{6}\d{4}\b")
# Rule 2: Account number starting with 990
dlp_acct_rule = re.compile(r"\b990\d{7}\b")

corpus = [
    ("999999XXXXXX0157", "Card"), # TP
    ("99000000157", "Acct"),      # TP
    ("Hello World", "Benign"),    # TN
    ("9999990000000157", "Benign"), # TN (unmasked, fails rule)
    ("9900 0000157", "Evasion")   # FN (spaces added)
]

tp, fp, fn, tn = 0, 0, 0, 0
for text, label in corpus:
    is_card = dlp_card_rule.search(text) is not None
    is_acct = dlp_acct_rule.search(text) is not None
    flagged = is_card or is_acct

    if flagged and label != "Benign": tp += 1
    elif flagged and label == "Benign": fp += 1
    elif not flagged and label == "Benign": tn += 1
    elif not flagged and label != "Benign": fn += 1

print(f"TP: {tp}, FP: {fp}, FN: {fn}, TN: {tn}")
print(f"Precision: {tp/(tp+fp) if (tp+fp)>0 else 0}, Recall: {tp/(tp+fn)}")
print("[PASS] DLP Evasion demonstrated: '9900 0000157' bypassed regex due to whitespace.")
```

### Q3(a) Account-Level Table & QI Classification
We aggregate 500 transactions into unique accounts. Multi-channel accounts are resolved by taking the most frequent terminal type.

```python
# Aggregate to account level
# Sensitive attribute: 1 if any transaction was DECLINED, else 0
df_clean['is_declined'] = (df_clean['TRANS_STAT'] == 'DECLINED').astype(int)

# Get mode for terminal type (tie-break: first alphabetically)
def get_mode(series):
    return series.value_counts().index[0] if not series.empty else "UNKNOWN"

acct_level = df_clean.groupby('ACCT_NO').agg(
    city=('ADDRESS_CITY', 'first'),
    month=('SWITCH_DATE', lambda x: str(x).zfill(6)[:6]), # Extract YYYYMM
    terminal_type=('TERMINAL_TYPE', get_mode),
    sensitive_declined=('is_declined', 'max')
).reset_index()

print(f"Account-level table created with {len(acct_level)} unique accounts.")
assert len(acct_level) < len(df_clean), "Aggregation failed to reduce records."
print("[PASS] Transaction data aggregated to person-level.")
```

```python
### Q3(b) Linkage Attack Simulation

def calculate_k(df, qis):
    grouped = df.groupby(qis).size().reset_index(name='k')
    return grouped

qis = ['city', 'month', 'terminal_type']
k_report = calculate_k(acct_level, qis)

unique_accounts = (k_report['k'] == 1).sum()
total_accounts = len(acct_level)
print(f"Unique identifiable accounts (k=1): {unique_accounts} out of {total_accounts} ({(unique_accounts/total_accounts)*100:.2f}%)")

# Add one more QI
qis_extra = ['city', 'month', 'terminal_type', 'sensitive_declined']
k_report_extra = calculate_k(acct_level, qis_extra)
unique_accounts_extra = (k_report_extra['k'] == 1).sum()
print(f"Unique identifiable accounts with extra QI: {unique_accounts_extra} ({(unique_accounts_extra/total_accounts)*100:.2f}%)")

assert unique_accounts_extra >= unique_accounts, "Adding QI should not decrease uniqueness!"
print("[PASS] Demonstrated increased risk with additional quasi-identifier.")
```

### Q3(c, d, e) k-Anonymity, l-Diversity, and t-Closeness
We implement generalisation for k>=5, check l-diversity (l>=2), and calculate t-closeness (TV distance <= 0.15).

```python
# Q3(c) Generalise city to region (simulated), month to quarter
def generalise_month(m):
    m = str(m).strip()
    # Check if the string is at least 6 characters long and the month part is numeric
    if len(m) >= 6 and m[4:6].isdigit():
        month_num = int(m[4:6])
        # Ensure it's a valid month (1 to 12)
        if 1 <= month_num <= 12:
            return m[:4] + "Q" + str((month_num - 1)//3 + 1)

    # Fallback for missing or invalid dates
    return "UNKNOWN_Q"

# Apply the generalisation
acct_level['gen_city'] = acct_level['city'].apply(lambda x: "REGION_A" if x in ["Durban", "Cape Town"] else "REGION_B")
acct_level['gen_month'] = acct_level['month'].apply(generalise_month)

qis_gen = ['gen_city', 'gen_month', 'terminal_type']
k_gen = calculate_k(acct_level, qis_gen)

# Suppress groups < 5
valid_groups = k_gen[k_gen['k'] >= 5]
suppressed_count = total_accounts - valid_groups['k'].sum()
print(f"k-Anonymity check: Min group size = {valid_groups['k'].min()}")
print(f"Suppressed accounts: {suppressed_count} ({(suppressed_count/total_accounts)*100:.2f}%)")

# EXECUTABLE PROOF (Adversarial test)
min_k = valid_groups['k'].min()
assert min_k >= 5, f"k-Anonymity failed! Found a group with size {min_k}."
print(f"[PASS] Proof: The minimum equivalence class size is {min_k} (k >= 5).")
```

```python
# Q3(d) l-Diversity
# Merge back to get sensitive attribute for valid groups
df_k = acct_level.merge(valid_groups[qis_gen], on=qis_gen, how='inner')
l_div = df_k.groupby(qis_gen)['sensitive_declined'].nunique().reset_index(name='l')
min_l = l_div['l'].min()
print(f"Minimum l-diversity: {min_l}")
if min_l < 2:
    print("[WARNING] l-diversity failed for some groups. Remediation: merge groups or suppress further.")
    # W3(d) Infeasibility witness: if all accounts in a group are declined, l=1. Must suppress.

# Q3(e) t-Closeness (Total Variation Distance for binary attribute)
global_mean = acct_level['sensitive_declined'].mean()
t_distance = 0
for _, group in df_k.groupby(qis_gen):
    local_mean = group['sensitive_declined'].mean()
    # TV distance for binary is |p - q|
    tv_dist = abs(local_mean - global_mean)
    t_distance = max(t_distance, tv_dist)

print(f"Max t-closeness (TV distance): {t_distance:.4f}")
if t_distance > 0.15:
    print("[WARNING] t-closeness failed (t > 0.15). Remediation required.")
else:
    print("[PASS] t-closeness (t <= 0.15) satisfied.")
```

### Q4 Synthetic Data Evaluation
We compare a Gaussian Copula generator (preserves pairwise dependencies) against a simple independent sampler. We assess utility and disclosure risk.

```python
import numpy as np
import pandas as pd
from scipy.stats import ks_2samp

# Original minimised data
df_orig = df_clean[['REQAMT', 'TERMINAL_TYPE', 'is_declined']].dropna()

# Generator A: Independent Sampler (Marginals only)
def gen_independent(df, n):
    syn = pd.DataFrame()
    for col in df.columns:
        syn[col] = np.random.choice(df[col], size=n, replace=True)
    return syn

# Generator B: Gaussian Copula (captures pairwise correlation)
# (For simplicity in Colab, we use a bivariate normal proxy for REQAMT and is_declined)
def gen_copula(df, n):
    # Standardise
    mean_req = df['REQAMT'].mean()
    std_req = df['REQAMT'].std()

    # Correlation
    corr = df[['REQAMT', 'is_declined']].corr().iloc[0,1]

    # Generate bivariate normal
    cov = [[1, corr], [corr, 1]]
    z = np.random.multivariate_normal([0, 0], cov, n)

    # Transform back
    syn_req = z[:, 0] * std_req + mean_req
    syn_declined = (z[:, 1] > 0).astype(int) # Simple threshold

    syn = pd.DataFrame({'REQAMT': syn_req, 'is_declined': syn_declined})
    syn['TERMINAL_TYPE'] = np.random.choice(df['TERMINAL_TYPE'], size=n)
    return syn

syn_indep = gen_independent(df_orig, len(df_orig))
syn_copula = gen_copula(df_orig, len(df_orig))

# Utility Metric: KS statistic for REQAMT (lower is better)
ks_indep = ks_2samp(df_orig['REQAMT'], syn_indep['REQAMT']).statistic
ks_copula = ks_2samp(df_orig['REQAMT'], syn_copula['REQAMT']).statistic

# Correlation preservation
orig_corr = df_orig['REQAMT'].corr(df_orig['is_declined'])
indep_corr = syn_indep['REQAMT'].corr(syn_indep['is_declined'])
copula_corr = syn_copula['REQAMT'].corr(syn_copula['is_declined'])

print(f"KS Distance (REQAMT): Independent={ks_indep:.4f}, Copula={ks_copula:.4f}")
print(f"Correlation (Original={orig_corr:.4f}): Independent={indep_corr:.4f}, Copula={copula_corr:.4f}")

# Disclosure Risk: Exact Overlap
overlap = len(pd.merge(df_orig, syn_copula, on=['REQAMT', 'TERMINAL_TYPE', 'is_declined']))
print(f"Exact record overlap: {overlap} rows")
assert overlap == 0, "Synthetic data leaked exact original records!"
print("[PASS] Utility vs. Risk trade-off assessed. No exact overlap.")
```

### Q5 Differential Privacy Mechanisms
We implement Laplace and Gaussian mechanisms for account-level COUNT and SUM, bounded by contribution limits. We maintain a privacy budget ledger.

```python
import numpy as np

# 5(a) Sensitivity and Clipping
# Clip each account's transaction sum to a max of 100,000
acct_sums = df_clean.groupby('ACCT_NO')['REQAMT'].sum().clip(upper=100000)
true_count = len(acct_sums)
true_sum = acct_sums.sum()

SENSITIVITY_COUNT = 1
SENSITIVITY_SUM = 100000 # Max contribution per account

# 5(b) Laplace Mechanism
def laplace_mechanism(true_value, sensitivity, epsilon):
    scale = sensitivity / epsilon
    noise = np.random.laplace(0, scale)
    return true_value + noise

# 5(c) Privacy Budget Ledger
EPSILON_TOTAL = 2.0
DELTA_TOTAL = 1e-5
ledger = {"epsilon": 0.0, "delta": 0.0}

def spend_budget(eps, delta=0.0):
    if ledger["epsilon"] + eps > EPSILON_TOTAL or ledger["delta"] + delta > DELTA_TOTAL:
        raise ValueError("BUDGET EXCEEDED: Query rejected safely.")
    ledger["epsilon"] += eps
    ledger["delta"] += delta
    print(f"Budget spent: ε={eps}, δ={delta}. Total: ε={ledger['epsilon']}, δ={ledger['delta']}")

# Release 1: Noisy Count (ε=0.5)
eps_count = 0.5
spend_budget(eps_count)
noisy_count = laplace_mechanism(true_count, SENSITIVITY_COUNT, eps_count)
print(f"Noisy Count: {noisy_count:.2f}")

# Release 2: Noisy Sum (ε=0.5)
eps_sum = 0.5
spend_budget(eps_sum)
noisy_sum = laplace_mechanism(true_sum, SENSITIVITY_SUM, eps_sum)
print(f"Noisy Sum: {noisy_sum:.2f}")

# Release 3: Noisy Mean (ε=1.0)
eps_mean = 1.0
spend_budget(eps_mean)
# Mean sensitivity is roughly (Sum_sens/count) + (Count_sens * Sum/count^2)
noisy_mean = noisy_sum / max(noisy_count, 1)
print(f"Noisy Mean: {noisy_mean:.2f}")

# Adversarial Test: Exceeding Budget
try:
    spend_budget(1.0) # Total would be 3.0 > 2.0
    assert False, "Budget check failed to stop query."
except ValueError as e:
    print(f"[PASS] {e}")
```

### 6(e) Comparative Architecture Decision for PETs

To allow three partner institutions to calculate a joint statistic without sharing raw records, we evaluate four Privacy-Enhancing Technologies (PETs):

| PET | Trust Assumption | Collusion Resistance | Key Custody | Computational Overhead | Practical Fit |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **Secure Aggregation** | Honest-majority | Resists up to \(t\) colluding parties | Threshold keys | Moderate | **Best fit** for 3 institutions. Handles dropouts via secret sharing. |
| **SMPC (General)** | Honest-majority | Resists \(t < n/2\) colluders | Distributed | High | Overkill for a simple sum; very complex to deploy. |
| **Additive HE** | Trusted key holder | Vulnerable if key holder colludes | Centralized | Very High | Impractical for 3 small institutions due to compute costs. |
| **TEE (e.g., Intel SGX)** | Trust in hardware vendor | Vulnerable to side-channel attacks | Vendor-managed | Low | Convenient, but requires hardware attestation and trust in the cloud provider. |

**Recommendation:**
A **Secure Aggregation** protocol using threshold secret sharing (e.g., Bonawitz et al.) combined with differential privacy (adding noise to the final aggregate) is the most practical fit. It tolerates the interruption of the smaller participant without compromising the privacy of the remaining institutions' raw records.
